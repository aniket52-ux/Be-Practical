Calculate the sum of even and odd numbers till N

```python
n = int(input("Enter N: "))
sum_even = 0
sum_odd = 0
for i in range(1,n+1):
  if i % 2 == 0:
    sum_even += i
  else:
    sum_odd += i
print("Even Sum", sum_even)
print("Odd Sum", sum_odd)
```
